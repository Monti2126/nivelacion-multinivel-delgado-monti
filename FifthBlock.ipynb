```python
def combatir(nombre_enemigo, vida_enemigo, ataque_enemigo,
             vida_heroe, ataque_heroe, defensa_heroe, max_vida):

    ronda = 1

    while vida_heroe > 0 and vida_enemigo > 0:

        dano_heroe = ataque_heroe
        vida_enemigo -= dano_heroe

        if vida_enemigo <= 0:
            print(f'{nombre_enemigo} derrotado')
            break

        dano_enemigo = ataque_enemigo - defensa_heroe

        if dano_enemigo <= 0:
            dano_enemigo = 1

        vida_heroe -= dano_enemigo

        print(f'Ronda {ronda} | Heroe: {vida_heroe} HP | {nombre_enemigo}: {vida_enemigo} HP')

        ronda += 1

    if vida_heroe > 0:
        vida_heroe += 20

        if vida_heroe > max_vida:
            vida_heroe = max_vida

    return vida_heroe


nombre_heroe = 'Guerrero'
nivel = 2
vida_heroe = 80
max_vida = 80
ataque_heroe = 18
defensa_heroe = 8

enemigos = [
    ['Goblin', 40, 8],
    ['Orco', 70, 14],
    ['Dragon', 120, 25]
]

enemigos_derrotados = 0

for enemigo in enemigos:

    nombre = enemigo[0]
    vida = enemigo[1]
    ataque = enemigo[2]

    print(f'\nCombate contra {nombre}')

    vida_heroe = combatir(
        nombre,
        vida,
        ataque,
        vida_heroe,
        ataque_heroe,
        defensa_heroe,
        max_vida
    )

    if vida_heroe <= 0:
        print('\nGAME OVER')
        break

    enemigos_derrotados += 1
    print(f'Vida restante del heroe: {vida_heroe}')

if enemigos_derrotados == 3 and vida_heroe > 0:
    print('\nVICTORIA')

print(f'\nEnemigos derrotados: {enemigos_derrotados}')
print(f'Vida final del heroe: {vida_heroe}')
```
